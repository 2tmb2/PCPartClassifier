```python
from pathlib import Path
from sklearn.dummy import DummyClassifier
from PIL import Image
import numpy as np
```

## Image To Matrix

```python
def image_to_matrix(image_path):
    img = Image.open(image_path)
    return np.array(img)
```

## Simple Baseline

```python
# load all image paths in dataset and associated true labels
image_paths = list(Path("laptop_parts").glob("*/*.jpg"))
true_labels = [path.parent.name for path in image_paths]
# initialize simple uniform classifier
model = DummyClassifier(strategy="uniform")
model.fit([np.ndarray(image_to_matrix(image_paths[0]).shape)] * len(image_paths), true_labels)

# test on a single input
predicted_label = model.predict([image_to_matrix(image_paths[0])])[0]
loss = np.log(25 + np.e)
if predicted_label == true_labels[0]:
    loss -= 1
print (f"Predicted Label: {predicted_label}\nTrue Label: {true_labels[0]}\nLoss: {loss}")
```

## Simple Baseline Training Error

```python
# accumulators
num_correct = 0
num_incorrect = 0
acc_loss = 0
ln_25_plus_e = np.log(25 + np.e) # for reuse

# determine loss from each one
for path, target in zip(image_paths, true_labels):
    # make prediction on current input
    pred = model.predict([image_to_matrix(path)])
    if pred == target:
        num_correct += 1
        acc_loss += ln_25_plus_e - 1
    else:
        num_incorrect += 1
        acc_loss += ln_25_plus_e

# print average loss over entire dataset, plus accuracy
print (f"Accuracy: {num_correct}/{num_correct + num_incorrect} ({num_correct / (num_correct + num_incorrect) * 100:.2f}%)")
print (f"Average Loss: {acc_loss / len(image_paths)}")
```
